```python
from getpass import getpass
import subprocess

token = getpass("Masukkan token GitHub: ")

try:
    push_url = REPO_URL.replace(
        "https://",
        f"https://{token}@",
        1
    )

    hasil = subprocess.run(
        ["git", "push", "-u", push_url, "main"],
        cwd=str(ROOT),
        capture_output=True,
        text=True
    )

    print(hasil.stdout)

    if hasil.returncode != 0:
        print("Terjadi kesalahan:")
        print(hasil.stderr)
    else:
        print("Proyek berhasil diunggah ke GitHub.")

finally:
    token = None
    push_url = None
```
